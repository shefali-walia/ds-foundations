```python
%%sql
CREATE TABLE aqi AS 
SELECT * FROM read_csv_auto('../week03-pandas/aqi_cleaned.csv')
```

**Q1.** Retrieve all columns for every record where `City = 'Delhi'`.

```python
%%sql
SELECT * FROM aqi WHERE City = 'Delhi';
```

**Q2.** Retrieve `City`, `Date`, and `AQI` for records where `City` is one of `'Delhi'`, `'Mumbai'`, or `'Bengaluru'` (use `IN`).

```python
%%sql
SELECT City, Date, AQI FROM aqi WHERE City IN ('Delhi','Bengaluru','Mumbai');
```

**Q3.** Retrieve all records for cities whose name starts with `'B'` (use `LIKE`).


```python
%%sql
SELECT * FROM aqi WHERE City LIKE 'B%';
```

**Q4.** Retrieve all records where `AQI` is greater than 300 AND `City = 'Delhi'`.

```python
%%sql
SELECT * FROM aqi WHERE AQI > 300 AND City = 'Delhi';
```

**Q5.** Find the overall average `AQI` across the whole dataset.

```python
%%sql
SELECT AVG(AQI) AS avgAQI FROM aqi;
```

**Q6.** Find the minimum and maximum `AQI` ever recorded, and which `City`/`Date` each occurred on (hint: you'll need this in two steps, or a subquery — min/max don't tell you *which row* by themselves).

```python
%%sql
SELECT City, Date, AQI FROM aqi WHERE AQI = (SELECT MIN(AQI) FROM aqi)
UNION ALL
SELECT City, Date, AQI FROM aqi WHERE AQI = (SELECT MAX(AQI) FROM aqi);
```

**Q7.** Count the number of distinct cities in the dataset.

```python
%%sql
SELECT COUNT(DISTINCT(City)) AS cnt FROM aqi;
```

**Q8.** Count how many records have a NULL `AQI` value.

```python
%%sql
SELECT COUNT(AQI) AS NullCnt FROM aqi WHERE AQI = Null;
```

**Q9.** Retrieve the 10 worst-AQI records overall (highest AQI, descending), showing `City`, `Date`, `AQI`.

```python
%%sql
SELECT City, Date, AQI FROM aqi ORDER BY AQI DESC;
```

**Q10.** Retrieve records 11–20 by AQI rank (i.e. skip the top 10, show the next 10) — use `LIMIT` with `OFFSET`.

```python
%%sql
SELECT City, Date, AQI FROM aqi ORDER BY AQI DESC LIMIT 10 OFFSET 10;
```
