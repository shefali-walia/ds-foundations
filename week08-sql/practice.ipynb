**Q1.** Retrieve all columns for every record where `City = 'Delhi'`.

```python
%%sql
SELECT * FROM aqi WHERE City = 'Delhi';
```

**Q2.** Retrieve `City`, `Date`, and `AQI` for records where `City` is one of `'Delhi'`, `'Mumbai'`, or `'Bengaluru'` (use `IN`).

```python
%%sql
SELECT City, Date, AQI FROM aqi WHERE City IN ('Delhi','Bengaluru','Mumbai');
```

**Q3.** Retrieve all records for cities whose name starts with `'B'` (use `LIKE`).


```python
%%sql
SELECT * FROM aqi WHERE City LIKE 'B%';
```

**Q4.** Retrieve all records where `AQI` is greater than 300 AND `City = 'Delhi'`.

```python
%%sql
SELECT * FROM aqi WHERE AQI > 300 AND City = 'Delhi';
```

**Q5.** Find the overall average `AQI` across the whole dataset.

```python
%%sql
SELECT AVG(AQI) AS avgAQI FROM aqi;
```

**Q6.** Find the minimum and maximum `AQI` ever recorded, and which `City`/`Date` each occurred on (hint: you'll need this in two steps, or a subquery — min/max don't tell you *which row* by themselves).

```python
%%sql
SELECT City, Date, AQI FROM aqi WHERE AQI = (SELECT MIN(AQI) FROM aqi)
UNION ALL
SELECT City, Date, AQI FROM aqi WHERE AQI = (SELECT MAX(AQI) FROM aqi);
```

**Q7.** Count the number of distinct cities in the dataset.

```python
%%sql
SELECT COUNT(DISTINCT(City)) AS cnt FROM aqi;
```

**Q8.** Count how many records have a NULL `AQI` value.

```python
%%sql
SELECT COUNT(AQI) AS NullCnt FROM aqi WHERE AQI = Null;
```

**Q9.** Retrieve the 10 worst-AQI records overall (highest AQI, descending), showing `City`, `Date`, `AQI`.

```python
%%sql
SELECT City, Date, AQI FROM aqi ORDER BY AQI DESC;
```

**Q10.** Retrieve records 11–20 by AQI rank (i.e. skip the top 10, show the next 10) — use `LIMIT` with `OFFSET`.

```python
%%sql
SELECT City, Date, AQI FROM aqi ORDER BY AQI DESC LIMIT 10 OFFSET 10;
```

**Q11.** Find the average `AQI` per `City`, sorted from worst to best.


```python
%%sql
SELECT City, AVG(AQI) AS avgAQI FROM aqi GROUP BY City ORDER BY avgAQI DESC;
```

**Q12.** Find cities where the average `AQI` is above 200 (use `HAVING`).


```python
%%sql
SELECT City, AVG(AQI) AS avgAQI FROM aqi GROUP BY City HAVING avgAQI > 200;
```

**Q13.** Count how many records fall into each `AQI_Bucket` category, sorted by count descending.

```python
%%sql
SELECT AQI_BUCKET, COUNT(AQI_BUCKET) AS cnt FROM aqi GROUP BY AQI_Bucket ORDER BY cnt DESC;
```

**Q14.** Find the average `PM2_5` and average `PM10` per city, but only for cities with more than 100 records (filter on `COUNT(*)` in `HAVING`).

```python
%%sql
SELECT City, AVG("PM2.5") AS avgPM25, AVG(PM10) AS avgPM10 FROM aqi GROUP BY City HAVING COUNT(*) >100;
```

**Q15.** Without using the existing `AQI_Bucket` column, write your own `CASE` expression that buckets `AQI` into `'Good' (<=50)`, `'Moderate' (<=100)`, `'Poor' (<=200)`, `'Severe' (>200)`. Alias it `custom_bucket`.

```python
%%sql
SELECT City, Date, 
CASE 
    WHEN AQI <=50 THEN 'Good'
    WHEN AQI <=100 THEN 'Moderate'
    WHEN AQI <=200 THEN 'Poor'
    ELSE 'Severe'
END AS custom_bucket FROM aqi;
```

**Q16.** Using the `CASE` from Q15 as a subquery, count how many records fall into each `custom_bucket`. Compare this against Q13's counts using the original `AQI_Bucket`.

```python
%%sql
SELECT b.custom_bucket, COUNT(*) AS cnt 
FROM (SELECT
CASE 
    WHEN AQI <=50 THEN 'Good'
    WHEN AQI <=100 THEN 'Moderate'
    WHEN AQI <=200 THEN 'Poor'
    ELSE 'Severe'
END AS custom_bucket FROM aqi) b GROUP BY custom_bucket;
```

**Q17.** Find all cities whose average `AQI` is above the *overall* average `AQI` for the whole dataset (you'll need a subquery in the `WHERE`/`HAVING` clause, not a hardcoded number).

```python
%%sql
SELECT DISTINCT(City) FROM aqi WHERE AQI > (SELECT AVG(AQI) FROM aqi);
```
